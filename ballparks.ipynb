```python
import geopandas as gpd
import pandas as pd
```

```python
states = gpd.read_file('ne_50m_admin_1_states_provinces_lakes (1)/ne_50m_admin_1_states_provinces_lakes.shp')
```

```python
states = states[states['iso_a2'].isin(['US', 'CA'])]
```

```python
states.to_file('states.geojson', driver='GeoJSON')
```
